# F1 2026 — Exploratory Data Analysis
Quick look at race results, driver trends, and constructor standings (2018–2025).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='darkgrid')

df = pd.read_csv('../data/raw/ergast_results_2018_2025.csv')
df.head()
```

```python
# Points per season by driver (top 10)
season_pts = df.groupby(['year','driver'])['points'].sum().reset_index()
top_drivers = season_pts.groupby('driver')['points'].sum().nlargest(10).index
pivot = season_pts[season_pts.driver.isin(top_drivers)].pivot(index='year', columns='driver', values='points')
pivot.plot(figsize=(12,5), marker='o')
plt.title('Season Points — Top 10 Drivers')
plt.tight_layout()
plt.savefig('../reports/figures/top_drivers_season_pts.png', dpi=150)
plt.show()
```

```python
# Grid vs finish position correlation
tmp = df.copy()
tmp['position'] = pd.to_numeric(tmp['position'], errors='coerce')
tmp = tmp.dropna(subset=['grid','position'])
plt.figure(figsize=(7,5))
sns.regplot(data=tmp, x='grid', y='position', scatter_kws={'alpha':0.2})
plt.title('Grid Position vs Finishing Position')
plt.tight_layout()
plt.savefig('../reports/figures/grid_vs_finish.png', dpi=150)
plt.show()
```
